# PySpark SQL Business Project — Automobile Analytics

**Input files:** `vehicles.csv`, `service_records.csv`

**Scenarios:** 21

This notebook uses Spark SQL for business analysis. All SQL is self-contained after the setup cell.

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import to_date, col

spark = SparkSession.builder.appName('PySparkSQLProject').getOrCreate()

from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('PySparkSQLProject').getOrCreate()

entities = spark.read.option('header', True).option('inferSchema', True).csv('vehicles.csv')
transactions = spark.read.option('header', True).option('inferSchema', True).csv('service_records.csv')

transactions = transactions.withColumn('service_date', to_date(col('service_date')))

entities.createOrReplaceTempView('entities')
transactions.createOrReplaceTempView('transactions')

print('Entities:', entities.count())
print('Transactions:', transactions.count())
entities.printSchema()
transactions.printSchema()
```

## Q1. Basic SELECT

Display sample transaction records.

```python
result = spark.sql("""
SELECT * FROM transactions LIMIT 10
""")
result.show(20, truncate=False)
```

## Q2. Column Selection

Select important business columns.

```python
result = spark.sql("""
SELECT vehicle_id, city, service_cost FROM transactions LIMIT 20
""")
result.show(20, truncate=False)
```

## Q3. WHERE Filter

Find transactions above the overall average.

```python
result = spark.sql("""
SELECT * FROM transactions WHERE service_cost > (SELECT AVG(service_cost) FROM transactions)
""")
result.show(20, truncate=False)
```

## Q4. BETWEEN Filter

Filter values between minimum and average.

```python
result = spark.sql("""
SELECT * FROM transactions WHERE service_cost BETWEEN (SELECT MIN(service_cost) FROM transactions) AND (SELECT AVG(service_cost) FROM transactions)
""")
result.show(20, truncate=False)
```

## Q5. IN Filter

Keep the three most frequent categories.

```python
result = spark.sql("""
SELECT * FROM transactions WHERE service_type IN (SELECT service_type FROM transactions GROUP BY service_type ORDER BY COUNT(*) DESC LIMIT 3)
""")
result.show(20, truncate=False)
```

## Q6. LIKE Search

Find entity records whose category/name contains the letter a.

```python
result = spark.sql("""
SELECT * FROM entities WHERE city LIKE '%a%'
""")
result.show(20, truncate=False)
```

## Q7. CASE WHEN

Create a business value segment.

```python
result = spark.sql("""
SELECT vehicle_id, service_cost,
CASE WHEN service_cost < (SELECT AVG(service_cost) FROM transactions) * 0.7 THEN 'Low'
     WHEN service_cost <= (SELECT AVG(service_cost) FROM transactions) * 1.3 THEN 'Medium'
     ELSE 'High' END AS value_segment
FROM transactions
""")
result.show(20, truncate=False)
```

## Q8. GROUP BY Count

Count records by category.

```python
result = spark.sql("""
SELECT service_type, COUNT(*) AS record_count FROM transactions GROUP BY service_type ORDER BY record_count DESC
""")
result.show(20, truncate=False)
```

## Q9. GROUP BY Aggregation

Calculate major KPIs by category.

```python
result = spark.sql("""
SELECT service_type, SUM(service_cost) AS total_value, AVG(service_cost) AS avg_value, MIN(service_cost) AS min_value, MAX(service_cost) AS max_value FROM transactions GROUP BY service_type ORDER BY total_value DESC
""")
result.show(20, truncate=False)
```

## Q10. HAVING

Show categories whose total is above the average category total.

```python
result = spark.sql("""
SELECT service_type, SUM(service_cost) AS total_value FROM transactions GROUP BY service_type HAVING SUM(service_cost) > (SELECT AVG(total_value) FROM (SELECT SUM(service_cost) AS total_value FROM transactions GROUP BY service_type)) ORDER BY total_value DESC
""")
result.show(20, truncate=False)
```

## Q11. INNER JOIN

Join entity and transaction data.

```python
result = spark.sql("""
SELECT e.*, t.service_date, t.service_cost, t.service_type
FROM entities e
INNER JOIN transactions t ON e.vehicle_id = t.vehicle_id
LIMIT 50
""")
result.show(20, truncate=False)
```

## Q12. JOIN Aggregation

Calculate joined business KPIs.

```python
result = spark.sql("""
SELECT e.city, COUNT(t.vehicle_id) AS transaction_count, SUM(t.service_cost) AS total_value
FROM entities e
JOIN transactions t ON e.vehicle_id=t.vehicle_id
GROUP BY e.city
ORDER BY total_value DESC
""")
result.show(20, truncate=False)
```

## Q13. LEFT JOIN

Find transaction coverage for every entity.

```python
result = spark.sql("""
SELECT e.vehicle_id, e.city, COUNT(t.vehicle_id) AS transaction_count
FROM entities e
LEFT JOIN transactions t ON e.vehicle_id=t.vehicle_id
GROUP BY e.vehicle_id, e.city
ORDER BY transaction_count DESC
""")
result.show(20, truncate=False)
```

## Q14. CTE Analysis

Calculate category contribution percentage.

```python
result = spark.sql("""
WITH category_sales AS (
    SELECT service_type, SUM(service_cost) AS total_value
    FROM transactions
    GROUP BY service_type
)
SELECT service_type, total_value,
       ROUND(total_value * 100.0 / SUM(total_value) OVER (), 2) AS contribution_pct
FROM category_sales
ORDER BY total_value DESC
""")
result.show(20, truncate=False)
```

## Q15. ROW_NUMBER

Rank each transaction inside its category.

```python
result = spark.sql("""
SELECT *,
ROW_NUMBER() OVER (PARTITION BY service_type ORDER BY service_cost DESC) AS row_num
FROM transactions
""")
result.show(20, truncate=False)
```

## Q16. DENSE_RANK

Create dense ranking inside categories.

```python
result = spark.sql("""
SELECT *,
DENSE_RANK() OVER (PARTITION BY service_type ORDER BY service_cost DESC) AS category_rank
FROM transactions
""")
result.show(20, truncate=False)
```

## Q17. LAG Comparison

Compare each value with the previous record.

```python
result = spark.sql("""
SELECT *,
LAG(service_cost) OVER (PARTITION BY service_type ORDER BY service_date) AS previous_value,
service_cost - LAG(service_cost) OVER (PARTITION BY service_type ORDER BY service_date) AS change_value
FROM transactions
""")
result.show(20, truncate=False)
```

## Q18. Running Total

Calculate running totals by category.

```python
result = spark.sql("""
SELECT *,
SUM(service_cost) OVER (
    PARTITION BY service_type
    ORDER BY service_date
    ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
) AS running_total
FROM transactions
""")
result.show(20, truncate=False)
```

## Q19. Monthly Trend

Analyze monthly business performance.

```python
result = spark.sql("""
SELECT YEAR(service_date) AS year_value,
MONTH(service_date) AS month_value,
SUM(service_cost) AS total_value
FROM transactions
GROUP BY YEAR(service_date), MONTH(service_date)
ORDER BY year_value, month_value
""")
result.show(20, truncate=False)
```

## Q20. Top 10

Find the top ten transactions.

```python
result = spark.sql("""
SELECT *
FROM (
    SELECT *, ROW_NUMBER() OVER (ORDER BY service_cost DESC) AS rn
    FROM transactions
) x
WHERE rn <= 10
ORDER BY service_cost DESC
""")
result.show(20, truncate=False)
```

## Q21. Final Business Report

Create the final management KPI report.

```python
result = spark.sql("""
WITH category_kpi AS (
    SELECT service_type,
           COUNT(*) AS records,
           SUM(service_cost) AS total_value,
           AVG(service_cost) AS average_value,
           MIN(service_cost) AS minimum_value,
           MAX(service_cost) AS maximum_value
    FROM transactions
    GROUP BY service_type
)
SELECT *,
       DENSE_RANK() OVER (ORDER BY total_value DESC) AS business_rank
FROM category_kpi
ORDER BY business_rank
""")
result.show(20, truncate=False)
```
