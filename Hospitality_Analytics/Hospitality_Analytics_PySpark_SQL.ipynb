# PySpark SQL Business Project — Hospitality Analytics

**Input files:** `hotels.csv`, `hotel_bookings.csv`

**Scenarios:** 21

This notebook uses Spark SQL for business analysis. All SQL is self-contained after the setup cell.

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import to_date, col

spark = SparkSession.builder.appName('PySparkSQLProject').getOrCreate()

from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('PySparkSQLProject').getOrCreate()

entities = spark.read.option('header', True).option('inferSchema', True).csv('hotels.csv')
transactions = spark.read.option('header', True).option('inferSchema', True).csv('hotel_bookings.csv')

transactions = transactions.withColumn('booking_date', to_date(col('booking_date')))

entities.createOrReplaceTempView('entities')
transactions.createOrReplaceTempView('transactions')

print('Entities:', entities.count())
print('Transactions:', transactions.count())
entities.printSchema()
transactions.printSchema()
```

## Q1. Basic SELECT

Display sample transaction records.

```python
result = spark.sql("""
SELECT * FROM transactions LIMIT 10
""")
result.show(20, truncate=False)
```

## Q2. Column Selection

Select important business columns.

```python
result = spark.sql("""
SELECT hotel_id, city, room_charge FROM transactions LIMIT 20
""")
result.show(20, truncate=False)
```

## Q3. WHERE Filter

Find transactions above the overall average.

```python
result = spark.sql("""
SELECT * FROM transactions WHERE room_charge > (SELECT AVG(room_charge) FROM transactions)
""")
result.show(20, truncate=False)
```

## Q4. BETWEEN Filter

Filter values between minimum and average.

```python
result = spark.sql("""
SELECT * FROM transactions WHERE room_charge BETWEEN (SELECT MIN(room_charge) FROM transactions) AND (SELECT AVG(room_charge) FROM transactions)
""")
result.show(20, truncate=False)
```

## Q5. IN Filter

Keep the three most frequent categories.

```python
result = spark.sql("""
SELECT * FROM transactions WHERE room_type IN (SELECT room_type FROM transactions GROUP BY room_type ORDER BY COUNT(*) DESC LIMIT 3)
""")
result.show(20, truncate=False)
```

## Q6. LIKE Search

Find entity records whose category/name contains the letter a.

```python
result = spark.sql("""
SELECT * FROM entities WHERE city LIKE '%a%'
""")
result.show(20, truncate=False)
```

## Q7. CASE WHEN

Create a business value segment.

```python
result = spark.sql("""
SELECT hotel_id, room_charge,
CASE WHEN room_charge < (SELECT AVG(room_charge) FROM transactions) * 0.7 THEN 'Low'
     WHEN room_charge <= (SELECT AVG(room_charge) FROM transactions) * 1.3 THEN 'Medium'
     ELSE 'High' END AS value_segment
FROM transactions
""")
result.show(20, truncate=False)
```

## Q8. GROUP BY Count

Count records by category.

```python
result = spark.sql("""
SELECT room_type, COUNT(*) AS record_count FROM transactions GROUP BY room_type ORDER BY record_count DESC
""")
result.show(20, truncate=False)
```

## Q9. GROUP BY Aggregation

Calculate major KPIs by category.

```python
result = spark.sql("""
SELECT room_type, SUM(room_charge) AS total_value, AVG(room_charge) AS avg_value, MIN(room_charge) AS min_value, MAX(room_charge) AS max_value FROM transactions GROUP BY room_type ORDER BY total_value DESC
""")
result.show(20, truncate=False)
```

## Q10. HAVING

Show categories whose total is above the average category total.

```python
result = spark.sql("""
SELECT room_type, SUM(room_charge) AS total_value FROM transactions GROUP BY room_type HAVING SUM(room_charge) > (SELECT AVG(total_value) FROM (SELECT SUM(room_charge) AS total_value FROM transactions GROUP BY room_type)) ORDER BY total_value DESC
""")
result.show(20, truncate=False)
```

## Q11. INNER JOIN

Join entity and transaction data.

```python
result = spark.sql("""
SELECT e.*, t.booking_date, t.room_charge, t.room_type
FROM entities e
INNER JOIN transactions t ON e.hotel_id = t.hotel_id
LIMIT 50
""")
result.show(20, truncate=False)
```

## Q12. JOIN Aggregation

Calculate joined business KPIs.

```python
result = spark.sql("""
SELECT e.city, COUNT(t.hotel_id) AS transaction_count, SUM(t.room_charge) AS total_value
FROM entities e
JOIN transactions t ON e.hotel_id=t.hotel_id
GROUP BY e.city
ORDER BY total_value DESC
""")
result.show(20, truncate=False)
```

## Q13. LEFT JOIN

Find transaction coverage for every entity.

```python
result = spark.sql("""
SELECT e.hotel_id, e.city, COUNT(t.hotel_id) AS transaction_count
FROM entities e
LEFT JOIN transactions t ON e.hotel_id=t.hotel_id
GROUP BY e.hotel_id, e.city
ORDER BY transaction_count DESC
""")
result.show(20, truncate=False)
```

## Q14. CTE Analysis

Calculate category contribution percentage.

```python
result = spark.sql("""
WITH category_sales AS (
    SELECT room_type, SUM(room_charge) AS total_value
    FROM transactions
    GROUP BY room_type
)
SELECT room_type, total_value,
       ROUND(total_value * 100.0 / SUM(total_value) OVER (), 2) AS contribution_pct
FROM category_sales
ORDER BY total_value DESC
""")
result.show(20, truncate=False)
```

## Q15. ROW_NUMBER

Rank each transaction inside its category.

```python
result = spark.sql("""
SELECT *,
ROW_NUMBER() OVER (PARTITION BY room_type ORDER BY room_charge DESC) AS row_num
FROM transactions
""")
result.show(20, truncate=False)
```

## Q16. DENSE_RANK

Create dense ranking inside categories.

```python
result = spark.sql("""
SELECT *,
DENSE_RANK() OVER (PARTITION BY room_type ORDER BY room_charge DESC) AS category_rank
FROM transactions
""")
result.show(20, truncate=False)
```

## Q17. LAG Comparison

Compare each value with the previous record.

```python
result = spark.sql("""
SELECT *,
LAG(room_charge) OVER (PARTITION BY room_type ORDER BY booking_date) AS previous_value,
room_charge - LAG(room_charge) OVER (PARTITION BY room_type ORDER BY booking_date) AS change_value
FROM transactions
""")
result.show(20, truncate=False)
```

## Q18. Running Total

Calculate running totals by category.

```python
result = spark.sql("""
SELECT *,
SUM(room_charge) OVER (
    PARTITION BY room_type
    ORDER BY booking_date
    ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
) AS running_total
FROM transactions
""")
result.show(20, truncate=False)
```

## Q19. Monthly Trend

Analyze monthly business performance.

```python
result = spark.sql("""
SELECT YEAR(booking_date) AS year_value,
MONTH(booking_date) AS month_value,
SUM(room_charge) AS total_value
FROM transactions
GROUP BY YEAR(booking_date), MONTH(booking_date)
ORDER BY year_value, month_value
""")
result.show(20, truncate=False)
```

## Q20. Top 10

Find the top ten transactions.

```python
result = spark.sql("""
SELECT *
FROM (
    SELECT *, ROW_NUMBER() OVER (ORDER BY room_charge DESC) AS rn
    FROM transactions
) x
WHERE rn <= 10
ORDER BY room_charge DESC
""")
result.show(20, truncate=False)
```

## Q21. Final Business Report

Create the final management KPI report.

```python
result = spark.sql("""
WITH category_kpi AS (
    SELECT room_type,
           COUNT(*) AS records,
           SUM(room_charge) AS total_value,
           AVG(room_charge) AS average_value,
           MIN(room_charge) AS minimum_value,
           MAX(room_charge) AS maximum_value
    FROM transactions
    GROUP BY room_type
)
SELECT *,
       DENSE_RANK() OVER (ORDER BY total_value DESC) AS business_rank
FROM category_kpi
ORDER BY business_rank
""")
result.show(20, truncate=False)
```
